```python
#Manipulating Data
import pandas as pd

sales = pd.Series(
    [5000,7000,None,8000,9000,10000,None],
    index = [
        'monday','tuesday','wednesday','thursday','friday','saturday','sunday']
                  )
print(sales)
```

```python
print(sales.min()) #to check minimum value
print(sales.max()) #maximum value
print(sales.mean()) #mean value
print(sales.sum()) #addition/sum
print(sales.var()) #variance
print(sales.std()) #standard deviation
print(sales.dtype) #to check thedata type
```

```python
#aggregation
sales['tuesday'] = 9500
print(sales)
sales['jan']=None
print(sales)
```

```python
#deleting data
sales=sales.drop('tuesday')
print(sales)
```

```python
print(sales + 2000)
print(sales*2)
```

```python
#Handling Duplicate data
import pandas as pd
data = {
    'EmpID':[101,102,103,104,105,106,107,108,109,110,110],
    'Name':['Ravi','PRIYA','Arun','sneha','Kiran','ANJALI','Vikram','Meena','RAHUL','sneha','sneha'],
    'City':['Hyderabad','hyderabad','CHENNAI','chennai','Bangalore','BANGALORE','Mumbai','mumbai','Delhi','Hyderabad','Hyderabad'],
    'Salary':[30000,40000,3500,4500,32000,50000,38000,42000,36000,45000,45000]
}
df = pd.DataFrame(data)
print(df)
```

```python
df.duplicated()
```

```python
df[df.duplicated()]
```

```python
df[df.duplicated(keep=False)]
```

```python
#counting the number of duplicates
df.duplicated().sum()
```

```python
#extracting duplictes with specified column
d=df[df.duplicated()]
total = d[['Name','Salary']].sum()
print(total)
```

```python
duplicate_rows = df[df.duplicated(keep=False)]
print(duplicate_rows)
```

```python
#extracting the duplicate from one required column(mainly primary key)
df['EmpID'].duplicated()
```

```python
f = df['EmpID'].duplicated(keep=False)
print(f)
```

```python
#dropping duplictes
df.drop_duplicates()
print(df)
```

```python
#it keeps first duplicate value
df.drop_duplicates(keep='first')
```

```python
#it keeps last duplicate value
df.drop_duplicates(keep='last')
```

```python
df.drop_duplicates(keep=False)
```
